# Horizontal and vertical lines

The datasets contain binary 28 × 28 images with lines 16–24 pixels long. Angles are sampled uniformly from −7° to +7° around horizontal or vertical. Horizontal lines are 1–2 pixels thick and vertical lines are 4–5 pixels thick; pixel steps may add a pixel where the line changes direction.


```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import torch

repo_root = Path.cwd().resolve()
if not (repo_root / "pyproject.toml").exists():
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root / "src"))

from domain_knowledge_analysis import utils
```

```python
config = utils.load_config(repo_root / "config" / "vae_lines.yaml")

horizontal = utils.create_dataset(config, "lines_dataset", train=True)
vertical = utils.create_dataset(config, "vertical_lines_dataset", train=True)
print(f"Horizontal: {len(horizontal)} images; vertical: {len(vertical)} images")
```

```python
generator = torch.Generator().manual_seed(123)
indices = torch.randperm(len(horizontal), generator=generator)[:8]

fig, axes = plt.subplots(2, len(indices), figsize=(14, 4))
for row, (name, dataset) in enumerate([("Horizontal", horizontal), ("Vertical", vertical)]):
    for column, index in enumerate(indices):
        axes[row, column].imshow(dataset[index.item()][0][0], cmap="gray", vmin=0, vmax=1, interpolation="nearest")
        axes[row, column].axis("off")
    axes[row, 0].set_title(name)
plt.tight_layout()
plt.show()
```

Training uses 4,000 images, validation uses 1,000, and the test set contains 1,000 separately generated images. These are random draws from a small discrete image space, so identical pixel patterns can occur across splits.
